```python
df.columns = df.columns.str.lower()
print("Updated columns:")
print(df.columns)
```

```python
df['click_through_rate(ctr)'] = (df['clicks'] / df['impressions'] * 100).round(2)
print(df['click_through_rate(ctr)'].head(5))
```

```python
# Clean 'acquisition_cost' by removing '$' and ',' then converting to float
df['acquisition_cost'] = df['acquisition_cost'].replace('[\\$,]', '', regex=True).astype(float)

# Calculate CPC
df['click_per_cost(cpc)'] = (df['acquisition_cost'] / df['clicks']).round(2)

print("First 5 rows of CPC:")
print(df['click_per_cost(cpc)'].head(5))
```

```python
# Calculate Revenue column: Cost + (Cost * ROI)
df['revenue'] = (df['acquisition_cost'] * (1 + df['roi'])).round(2)

print("First 5 rows with the simplified Revenue column:")
display(df[['campaign_id', 'acquisition_cost', 'roi', 'revenue']].head())
```

```python
# Calculate ROAS (Fixed typo and ensured lowercase consistency)
df['return_on_ad_spend(roas)'] = (df['revenue'] / df['acquisition_cost']).round(2)
display(df[['campaign_id', 'revenue', 'acquisition_cost', 'return_on_ad_spend(roas)']].head(3))
```

```python
import matplotlib.pyplot as plt
```

```python
#average clicks per campaign type
avg_clicks = df.groupby('campaign_type')['clicks'].mean().sort_values()

#line chart
plt.figure(figsize=(5, 4))
plt.plot(avg_clicks.index, avg_clicks.values, marker='o', linestyle='-', color='teal', linewidth=2)

#Add labels and title
plt.title('Average Clicks by Campaign Type', fontsize=12)
plt.xlabel('Campaign Type', fontsize=10)
plt.ylabel('Average Clicks', fontsize=10)
plt.grid(axis='y', linestyle='--', alpha=0.7)

#Display the plot
plt.tight_layout()
plt.show()

#Show the raw averages for precision
print("Average Clicks per Campaign Type:")
print(avg_clicks)
```

```python
# Calculate average ROI per campaign type
avg_roi = df.groupby('campaign_type')['roi'].mean().sort_values()

#simple line plot
plt.figure(figsize=(4, 3))
plt.plot(avg_roi.index, avg_roi.values, marker='o', linestyle='-', color='#d62728', linewidth=2)

# Add labels and title
plt.title('Average ROI by Campaign Type (Matplotlib Only)', fontsize=12)
plt.xlabel('Campaign Type', fontsize=12)
plt.ylabel('Mean ROI', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.6)

# Display the plot
plt.show()
```

```python
# Calculate total acquisition spend per channel
channel_spend = df.groupby('channel_used')['acquisition_cost'].sum().sort_values()

#horizontal bar chart
plt.figure(figsize=(5,4))
plt.barh(channel_spend.index, channel_spend.values, color='skyblue', edgecolor='black')

# Add labels and title
plt.title('Total Acquisition Spend by Channel', fontsize=12)
plt.xlabel('Total Spend ($)', fontsize=7)
plt.ylabel('Channel', fontsize=7)
plt.grid(axis='x', linestyle='--', alpha=0.7)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(4, 3))
plt.hist(df['conversion_rate'], bins=20, edgecolor='black')
plt.title('Distribution of Conversion Rate')
plt.xlabel('Conversion Rate')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```
